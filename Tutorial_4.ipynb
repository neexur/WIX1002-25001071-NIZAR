# Question 1
1) This data is not a tidy data
2) The rule violated: Each variable must form a **column**

```python
import pandas as pd

data_q1 = {
    "Student": ["Ali", "Sarah", "John"],
    "Math": [75, 88, 65],
    "Science": [80, 91, 70],
    "English": [72, 85, 68]
}

df_q1 = pd.DataFrame(data_q1)
print(df_q1)


df_tidy_q1 = df_q1.melt(id_vars=["Student"], var_name="Subject", value_name="Score")
print(df_tidy_q1)
```

```python
import matplotlib.pyplot as plt

data_q3 = {
  "Student": ["A", "B", "C", "D", "E", "F"],
  "Study_Hours": [2, 5, 3, 7, 4, 8],
  "Score": [55, 72, 60, 85, 68, 90]
}

df_q3 = pd.DataFrame(data_q3)

print(f"1. Average study time: {df_q3['Study_Hours'].mean():.2f} hours")
print(f"2. Average score: {df_q3['Score'].mean():.2f}")

highest_score_student = df_q3.loc[df_q3['Score'].idxmax(), 'Student']
print(f"3. Student with highest score: {highest_score_student}")

most_studied_student = df_q3.loc[df_q3['Study_Hours'].idxmax(), 'Student']
print(f"4. Student who studied the most: {most_studied_student}")

# 3. Create Scatter Plot
plt.figure(figsize=(6, 4))
plt.scatter(df_q3['Study_Hours'], df_q3['Score'], color='blue', edgecolor='k', s=100, zorder=3)

# Annotate each point with the student's name
for i, txt in enumerate(df_q3['Student']):
    plt.annotate(txt, (df_q3['Study_Hours'].iloc[i] + 0.15, df_q3['Score'].iloc[i] - 0.5), fontsize=10)

plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(1, 9)
plt.ylim(50, 95)
plt.show()
```
